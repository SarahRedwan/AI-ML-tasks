```python
import pandas as pd
import numpy as np

df = pd.read_csv("students.csv")
print(df)
```

```python

def clean_data(df):
    df = df.copy()

    df["name"] = df["name"].astype("string").str.strip()

    df["score"] = pd.to_numeric(df["score"], errors="coerce")

    df = df.dropna(subset=["name", "score"])

    df = df[df["score"].between(0, 100)]

    return df.reset_index(drop=True)
```

```python

cleaned_df = clean_data(df)

print(cleaned_df)
```

```python

def analyze_scores(cleaned_df):
    # Convert the score column into a NumPy array
    scores = cleaned_df["score"].to_numpy(dtype=float)

    # Calculate statistics
    average = np.mean(scores)
    highest = np.max(scores)
    lowest = np.min(scores)

    # Count students who passed
    passed = np.sum(scores >= 50)

    # Calculate pass rate
    pass_rate = (passed / len(scores)) * 100

    # Return the results
    return {
        "total_students": len(scores),
        "average": average,
        "highest": highest,
        "lowest": lowest,
        "passed": int(passed),
        "pass_rate": pass_rate
    }
```

```python

results = analyze_scores(cleaned_df)

print(results)
```

```python

def display_report(cleaned_df, results):
    print("STUDENT SCORE ANALYZER")
    print("=" * 30)

    print("\nCleaned student records:")
    print(cleaned_df.to_string(index=False))

    print("\nSummary statistics:")
    print(f"Total students: {results['total_students']}")
    print(f"Average score: {results['average']:.2f}")
    print(f"Highest score: {results['highest']:.2f}")
    print(f"Lowest score: {results['lowest']:.2f}")
    print(f"Students passed: {results['passed']}")
    print(f"Pass rate: {results['pass_rate']:.2f}%")
```

```python

display_report(cleaned_df, results)

```
